```python
import glob
import pickle

from experiments.end_to_end.proof_node import Status

dir = '../runs/end_to_end/reprover-v1-novel-no-split-2/2024_05_22/15_07_08/traces/0/'
```

```python
files = glob.glob(dir + '*')
```

```python
from tqdm import tqdm

proven_explored = []
unproven_explored = []
for file in tqdm(files):
    try:
        trace = pickle.load(open(file, 'rb'))
        if trace.proof:
            # proven_explored.append(len([node for node in trace.nodes.values() if node.is_explored]))
            proven_explored.append(sum([t.time for t in trace.trace]) / len(trace.trace))
        elif trace.tree.status == Status.OPEN:
            # unproven_explored.append(len([node for node in trace.nodes.values() if node.is_explored]))
            # unproven_explored.append(len(trace.trace))
            unproven_explored.append(sum([t.time for t in trace.trace]) / len(trace.trace))
    except: 
        pass
```

```python
test_trace = None

for file in files:
    try:
        trace = pickle.load(open(file, 'rb'))
        if trace.tree.status == Status.OPEN:
            if len([node for node in trace.nodes.values() if node.is_explored]) < 50:
                test_trace = trace
                break
    except:
        pass
                
            
```
